```python
# Q1. Pass or Fail counter.

marks = [45, 32, 78, 39, 90, 28, 55, 40, 67, 12]
passed = 0
failed = 0

for mark in marks:
    if mark >= 40:
        print(mark, "Pass")
        passed += 1
    else:
        print(mark, "Fail")
        failed += 1

# List comprehension for passing marks
passing_marks = [mark for mark in marks if mark >= 40]

print("Total Passed:", passed)
print("Total Failed:", failed)
print("Passing Marks:", passing_marks)
```

```python
# Q2. Safe Average Calculator.
def calculate_average(numbers: list) -> float:
    """Returns the average of a list of numbers."""
    return sum(numbers) / len(numbers)

# Normal list
try:
    numbers = [10, 20, 30]
    result = calculate_average(numbers)
    print("Average:", result)
except ZeroDivisionError:
    print("Cannot calculate average of an empty list.")

# Empty list
try:
    numbers = []
    result = calculate_average(numbers)
    print("Average:", result)
except ZeroDivisionError:
    print("Cannot calculate average of an empty list.")
```

```python
# Q3. Daily Step Tracker

import random
import math

steps = []

for day in range(1, 8):
    daily_steps = random.randint(2000, 12000)
    steps.append(daily_steps)
    print("Day", day, ":", daily_steps, "steps")

highest = max(steps)
lowest = min(steps)

average = sum(steps) / len(steps)
average = math.ceil(average)

print("Highest steps:", highest)
print("Lowest steps:", lowest)
print("Average steps:", average)
```

```python
# Q4. Student Class

class Student:
    def __init__(self, name: str, marks: list):
        self.name = name
        self.marks = marks

    def average(self) -> float:
        """Returns the average marks of the student."""
        return sum(self.marks) / len(self.marks)

    def __str__(self):
        return f"{self.name}: average {self.average():.1f}"

student1 = Student("Yash", [80, 75, 90])
student2 = Student("Sakshi", [85, 90, 88])
student3 = Student("Rohan", [70, 65, 78])

students = [student1, student2, student3]

for student in students:
    print(student)

highest_student = max(students, key=lambda student: student.average())

print("Highest average:", highest_student.name)
```

```python
# Q5. Import and use Module.

import math
radius = 5

# Calculate the area of the circle
area = math.pi * math.pow(radius, 2)

# Calculate the circumference of the circle
circumference = 2 * math.pi * radius

# Round the area to 2 decimal places
rounded_area = round(area, 2)

print("Radius:", radius)
print("Area:", rounded_area)
print("Circumference:", round(circumference, 2))
```
